
**BMI Calculator using ipywidgets**

This version of the BMI calculator uses ipywidgets to create an interactive interface that can run directly within Google Colab. The core calculation logic remains the same, but the UI elements and event handling are adapted for the Colab environment.

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- Core Calculation Functions (from original code) ---

def calculate_bmi(weight, height):
    """
    Calculates the Body Mass Index (BMI) based on weight and height.
    Returns the calculated BMI value.
    """
    # Ensure height is not zero to prevent division by zero error
    if height == 0:
        return 0.0 # Or raise an error
    bmi = weight / (height ** 2)
    return bmi

def get_bmi_category(bmi):
    """
    Determines the BMI category based on the BMI value.
    Returns the BMI category as a string.
    """
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 25:
        return "Normal Weight"
    elif 25 <= bmi < 30:
        return "Overweight"
    else:
        return "Obese"

def get_weight_range(height):
    """
    Provides a suggested weight range based on the height.
    Returns a tuple containing the lower and upper weight limits.
    """
    lower_limit = 18.5 * (height ** 2)
    upper_limit = 24.9 * (height ** 2)
    return lower_limit, upper_limit

def get_height_range(weight):
    """
    Provides a suggested height range based on the weight.
    Returns a tuple containing the lower and upper height limits.
    """
    # Ensure weight is not zero for calculation
    if weight == 0:
        return (0.0, 0.0) # Or handle appropriately
    lower_limit = (weight / 24.9) ** 0.5
    upper_limit = (weight / 18.5) ** 0.5
    return lower_limit, upper_limit

# --- ipywidgets UI Elements ---

# Input widgets
weight_input = widgets.FloatText(
    value=70.0,
    description='Weight:',
    min=0.1
)
weight_unit_dropdown = widgets.Dropdown(
    options=['kgs', 'lbs'],
    value='kgs',
    description='Unit:',
)

height_input = widgets.FloatText(
    value=1.75,
    description='Height:',
    min=0.1
)
height_unit_dropdown = widgets.Dropdown(
    options=['meters', 'feet'],
    value='meters',
    description='Unit:',
)

calculate_button = widgets.Button(
    description='Calculate BMI',
    button_style='success',
    tooltip='Click to calculate BMI'
)

# Output widget to display results
output_area = widgets.Output()

# --- Calculation and Display Logic for ipywidgets ---

def on_calculate_button_clicked(b):
    with output_area:
        clear_output(wait=True)
        try:
            weight = weight_input.value
            height = height_input.value

            if weight <= 0 or height <= 0:
                print("Please enter positive values for weight and height.")
                return

            if weight_unit_dropdown.value == "lbs":
                weight = weight * 0.453592  # Convert lbs to kg

            if height_unit_dropdown.value == "feet":
                height = height * 0.3048  # Convert feet to meters

            bmi = calculate_bmi(weight, height)
            bmi_category = get_bmi_category(bmi)

            weight_range = get_weight_range(height)
            height_range = get_height_range(weight)

            print(f"BMI: {bmi:.2f}")
            print(f"Category: {bmi_category}")
            print(f"Suggested Weight Range: {weight_range[0]:.2f} - {weight_range[1]:.2f} kg")
            print(f"Suggested Height Range: {height_range[0]:.2f} - {height_range[1]:.2f} meters")
        except ValueError:
            print("Please enter valid numerical values.")
        except Exception as e:
            print(f"An error occurred: {e}")

# Attach the event handler to the button
calculate_button.on_click(on_calculate_button_clicked)

# --- Layout and Display ---

# Arrange input widgets
input_layout = widgets.VBox([
    widgets.HBox([weight_input, weight_unit_dropdown]),
    widgets.HBox([height_input, height_unit_dropdown])
])

# Combine all UI elements
ui = widgets.VBox([
    input_layout,
    calculate_button,
    output_area
])

display(ui)
```
